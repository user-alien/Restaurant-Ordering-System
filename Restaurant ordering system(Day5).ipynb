```python
#Restaurant Ordering System

import csv


class Restaurant:

    def __init__(self):
        # Dictionary + Tuple
        self.menu = {
            1: ("Burger", 100),
            2: ("Pizza", 200),
            3: ("Momos", 80),
            4: ("Fried Rice", 120),
            5: ("Chicken Roll", 90),
            6: ("Cold Drink", 40)
        }

        # List
        self.order = []

        # Set
        self.categories = {"Fast Food", "Main Course", "Drinks"}

    # Display menu
    def show_menu(self):
        print("\n----- RESTAURANT MENU -----")

        for number, item in self.menu.items():
            print(number, item[0], "₹", item[1])

    # Add food item
    def add_item(self):

        self.show_menu()

        try:
            choice = int(input("\nEnter item number: "))

            if choice in self.menu:

                quantity = int(input("Enter quantity: "))

                if quantity > 0:

                    name = self.menu[choice][0]
                    price = self.menu[choice][1]

                    self.order.append(
                        (name, price, quantity)
                    )

                    print("Item added successfully!")

                else:
                    print("Quantity must be greater than 0.")

            else:
                print("Invalid item number.")

        except ValueError:
            print("Please enter a valid number.")

    # Show order
    def show_order(self):

        if len(self.order) == 0:
            print("\nYour order is empty.")
            return

        print("\n----- YOUR ORDER -----")

        total = 0

        for item in self.order:

            name, price, quantity = item

            amount = price * quantity

            print(name, "x", quantity, "=", "₹", amount)

            total = total + amount

        print("----------------------")
        print("Total = ₹", total)

    # Calculate bill
    def bill(self):

        if len(self.order) == 0:
            print("\nNo items in your order.")
            return

        total = 0

        for item in self.order:

            name, price, quantity = item

            total = total + (price * quantity)

        # Discount
        if total >= 500:
            discount = total * 0.10
        else:
            discount = 0

        final_amount = total - discount

        print("\n----- BILL -----")
        print("Total     : ₹", total)
        print("Discount  : ₹", discount)
        print("Final Bill: ₹", final_amount)

    # Save order in CSV file
    def save_order(self):

        if len(self.order) == 0:
            print("\nNo order to save.")
            return

        try:

            with open("orders.csv", "w", newline="") as file:

                writer = csv.writer(file)

                writer.writerow(
                    ["Food", "Price", "Quantity", "Amount"]
                )

                for item in self.order:

                    name, price, quantity = item

                    amount = price * quantity

                    writer.writerow(
                        [name, price, quantity, amount]
                    )

            print("Order saved in orders.csv")

        except Exception:
            print("Error while saving file.")


# Create object
restaurant = Restaurant()


# Main menu
while True:

    print("\n==========================")
    print("   RESTAURANT SYSTEM")
    print("==========================")

    print("1. Show Menu")
    print("2. Add Food")
    print("3. Show Order")
    print("4. Calculate Bill")
    print("5. Save Order")
    print("6. Exit")

    try:

        choice = int(input("\nEnter your choice: "))

        if choice == 1:

            restaurant.show_menu()

        elif choice == 2:

            restaurant.add_item()

        elif choice == 3:

            restaurant.show_order()

        elif choice == 4:

            restaurant.bill()

        elif choice == 5:

            restaurant.save_order()

        elif choice == 6:

            print("\nThank you for visiting!")
            break

        else:

            print("Invalid choice! Choose 1-6.")

    except ValueError:

        print("Invalid input! Please enter a number.")


```
